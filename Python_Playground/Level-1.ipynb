9. Using the same lists, calculate the Mean Absolute Error (MAE).

Formula:

MAE = sum of absolute errors / number of predictions

```python
actual = [100, 200, 300, 400, 500]
predicted = [110, 190, 290, 420, 480]
sum=0
for i in range (len(actual)):
    abs_error = abs(actual[i]-predicted[i])
    print(abs_error)
    sum+=abs_error
mean_abs_err = sum/len(predicted)
print(f"MAE:",abs(mean_abs_err))

```

19. Given:

Calculate:

absolute error for each prediction
total error
MAE

using only loops.

```python
predicted = [10, 20, 30, 40, 50]
actual = [12, 18, 33, 39, 55]
total_error=0
for i in range(len(actual)):
    error=actual[i]-predicted[i]
    print(abs(error))
    total_error+=abs(error)
print(f"Total Error:",total_error)
mae = total_error/len(predicted)
print(f"MAE:",mae)
```
