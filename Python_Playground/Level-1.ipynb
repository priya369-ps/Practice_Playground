12. Take a person's age and print:
"Child" if below 13
"Teenager" if 13–19
"Adult" if 20 or above

```python
age = int(input("enter person's age:"))

if age <13:
    print("Child")
elif (age>=13) & (age<=19):
    print("Teenager")
else:
    print("Adult")
```

13. Take marks and assign grades:
90–100 → A
80–89 → B
70–79 → C
60–69 → D
Below 60 → F

```python
marks = float(input("Enter marks:"))

if (marks>=90) & (marks<=100):
    print(f"Grade: A")
elif marks>100:
    print("Scam/Paper leaked/great work if you did hardwork")
elif (marks>=80) & (marks<=89):
    print(f"Grade: B")
elif (marks>=70) & (marks<=79):
    print(f"Grade: C")
elif (marks>=60) & (marks<=69):
    print(f"Grade: D")
else:
    print("Fail")
```

15. Write a program to find the largest of three numbers.

```python
a = int(input("Enter a number:"))
b = int(input("Enter a number:"))
c = int(input("Enter a number:"))


if (a>b) & (a>c):
    print(f"A is bigger")
elif (b>a) & (b>c):
    print(f"B is bigger")
else:
    print(f"C is bigger")
```

16. Check whether a year is a leap year.

```python
year = int(input("enter a year:"))

if year%400 ==0:
    print("Given year is leap year")
elif year%4==0:
    print("Given year is leap year")
elif year%100 == 0:
    print("Given year is not a leap year")
else:
    print("Given year is not a leap year")
    
```
