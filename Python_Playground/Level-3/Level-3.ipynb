```python
filepath = "sample.csv"
def read_csv_manual(filepath):
    rows = []
    errors = []

    try:
        with open(filepath, "r") as f:
            header = f.readline().strip().split(",")
            expected_cols = len(header)

            for line_num, line in enumerate(f, start=2):  # start=2 since header is line 1
                line = line.strip()
                if not line:
                    continue  # skip blank lines

                fields = line.split(",")

                try:
                    if len(fields) != expected_cols:
                        raise ValueError(
                            f"Expected {expected_cols} columns, got {len(fields)}"
                        )
                    row = dict(zip(header, fields))
                    rows.append(row)

                except ValueError as e:
                    errors.append((line_num, str(e)))
                    continue  # skip bad row, keep processing

    except FileNotFoundError:
        print(f"File not found: {filepath}")
        return [], []

    return rows, errors


if __name__ == "__main__":
    data, bad_rows = read_csv_manual("sample.csv")
    print(f"Parsed {len(data)} rows successfully.")
    if bad_rows:
        print(f"{len(bad_rows)} bad rows skipped:")
        for line_num, msg in bad_rows:
            print(f"  Line {line_num}: {msg}")
```
